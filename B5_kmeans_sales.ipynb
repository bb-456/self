# B5: K-Means clustering on sales_data_sample.csv, number of clusters by the elbow method



## Import libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

## Step 1: Load and scale numeric features

```python
df = pd.read_csv("sales_data_sample.csv", encoding="latin1")
X = df[["QUANTITYORDERED", "PRICEEACH", "SALES", "MSRP"]]
X = StandardScaler().fit_transform(X)
```

## Step 2: Elbow method

```python
wcss = []
for k in range(1, 11):
    wcss.append(KMeans(k).fit(X).inertia_)
plt.plot(range(1, 11), wcss)
```

## Step 3: K-Means with k = 3 (the elbow)

```python
df["Cluster"] = KMeans(3).fit_predict(X)
plt.scatter(df["QUANTITYORDERED"], df["SALES"], c=df["Cluster"])
df["Cluster"].value_counts()
```
