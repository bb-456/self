# B4: Gradient Descent — Local minimum of y = (x + 3)^2 starting from x = 2


```python
import matplotlib.pyplot as plt

x = 2
learning_rate = 0.1

for i in range(100):
    gradient = 2 * (x + 3)   # slope dy/dx
    x = x - learning_rate * gradient   # IMPORTANT FORMULA  x_new = x_old - (learning_rate * gradient)
    plt.scatter(x, (x + 3) ** 2)

print("Local minimum at x =", round(x), ", y =", round((x + 3) ** 2))
plt.show()
```
